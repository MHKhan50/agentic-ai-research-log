# MemTrust-Lite dataset quick look

All data are synthetic. Results measure the simulator, not real LLM agents.

```python
import pandas as pd
from pathlib import Path
def locate_csv(name):
    candidates = [Path(name), Path('../data') / name, Path('../outputs') / name]
    kaggle_root = Path('/kaggle/input')
    if kaggle_root.exists():
        candidates.extend(kaggle_root.glob('**/' + name))
    for candidate in candidates:
        if candidate.exists():
            return candidate
    raise FileNotFoundError(name + ' not found; attach the dataset or run the project first.')
scenarios = pd.read_csv(locate_csv('scenarios.csv'))
results = pd.read_csv(locate_csv('results.csv'))
summary = pd.read_csv(locate_csv('summary.csv'))
print('scenarios:', scenarios.shape, 'results:', results.shape)
summary
```

```python
summary.pivot(index='poisoning_rate', columns='policy', values='attack_success_rate_on_poisoned')
```

## Interpretation reminder
Attack success is conditional on a poisoned memory being present. At 0% injection, the value is undefined (`NA`). Do not interpret simulator rates as real-world agent attack rates.
